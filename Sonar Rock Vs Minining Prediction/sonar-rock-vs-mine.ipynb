```python
import pandas as pd

# Header ko ignore karo, column names khud bana do
column_names = [str(i) for i in range(60)] + ['Label']  # 60 features + 1 label

# CSV load karo
solar_data = pd.read_csv('Copy of sonar data.csv', header=None, names=column_names)

print(solar_data.head())
print(solar_data['Label'].value_counts())
```

```python
solar_data
```

```python
solar_data.groupby('Label').mean()
```

```python
solar_data.describe()
```

### Train Test Split

```python
X=solar_data.drop(columns='Label',axis=1)
```

```python
X
```

```python
df=solar_data.copy()
```

```python
df
```

```python
y=df['Label']
```

```python
y
```

```python
print(X.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,stratify=y,random_state=1)
```

```python
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

### Build and Evaluate Model

```python
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression()
```

```python
model.fit(X_train,y_train)
```

```python
prediction=model.predict(X_train)
prediction
```

```python
from sklearn.metrics import accuracy_score
acc_score=accuracy_score(y_train,prediction)
print(acc_score)
```

```python
test_prediction = model.predict(X_test)
print(accuracy_score(test_prediction,y_test))
```
