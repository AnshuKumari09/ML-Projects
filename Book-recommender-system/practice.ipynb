You want to build a Collaborative Filtering Based Recommender System, and first step is:
“Find only those users who are active enough (have rated more than 200 books), so that recommendations become reliable.”

```python
ratings_with_name = ratings.merge(books,on='ISBN')
```

```python
ratings_with_name.groupby('User-ID')
```

```python
#Count ratings per user
ratings_with_name.groupby('User-ID').count()['Book-Rating']
```

```python
#filter
ratings_with_name.groupby('User-ID').count()['Book-Rating']>200
```

Keep only True users -> x[x] 👉 only True values remain 🧠 Why >200 threshold?
Get User IDs
.index 👉 final list of “active users”
Because:
low ratings users = noisy data
high ratings users = stable preferences
better similarity calculation

agar koi user sirf 2-3 books rate karta hai, uski preference reliable nahi hogi. Toh sirf woh users lo jinki "reading history" strong hai — 200+ books rate ki ho. x ek True/False Series hai. x[x] sirf True wale rakhta hai. .index se User-IDs milti hain.

| User | Likes                |
| ---- | -------------------- |
| A    | Harry Potter, LOTR   |
| B    | Harry Potter, Hobbit |
| C    | LOTR, Hobbit      

👉 If A and B are similar → recommend B’s books to A   |


```python
x = ratings_with_name.groupby('User-ID').count()['Book-Rating'] > 200
active_users = x[x].index
```

# FINAL STEP: Pivot Table (CORE IDEA)

BEFORE:

User1 → BookA → 5
User2 → BookA → 0
User3 → BookA 4 →

AFTER:

            User1  User2  User3
BookA         5      0      4
BookB         0      3     0 4

```python
final_ratings = filtered_rating[filtered_rating['Book-Title'].isin(famous_books)]
```

```python
pt = final_ratings.pivot_table(
    index='Book-Title',
    columns='User-ID',
    values='Book-Rating'
)
```

WHY pivot table?

Because collaborative filtering needs:

“Item-user matrix”

So we can compare:

Book vs Book similarity
User pattern similarity

```python
pt.head()
```

```python
pt.fillna(0, inplace=True)
```

# working

Users ka kaam tha:

| Book \ User | U1 | U2 | U3 | U4 |

| Book B      | 0  | 4  | 0  | 2  |

cosine_similarity(pt) kya karta hai?
cosine_similarity(pt)

👉 Har row (book) ko ek vector maanta hai

Book A = [5, 0, 3, 0]
Book B = [0, 4, 0, 2]
Book C = [1, 1, 05], 

Step 2: Pairwise comparison

Cosine similarity har pair compare karta hai:

Book A vs Book B
Book A vs Book C
Book B vs Book C

Step 3: Formula behind scenes

cos(θ)=A⋅B/∥A∥∥B∥

If two books:
same users rate same way → similarity HIGH (≈1)
different users → similarity LOW (≈0)
📊 4. Output: similarity_scores
similarity_scores = cosine_similarity(pt)


Why this is used in recommendation system?

When user likes Book A: 👉 system checks:
similarity_scores[A] → top similar books 👉 and recommends them5]

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity_scores = cosine_similarity(pt)
similarity_scores.shape
```

```python
def recommend(book_name):
    # index fetch
    index = np.where(pt.index==book_name)[0][0]
    similar_items = sorted(list(enumerate(similarity_scores[index])),key=lambda x:x[1],reverse=True)[1:5] #👉 index + similarity pair
    
    data = []
    for i in similar_items:
        item = []
        temp_df = books[books['Book-Title'] == pt.index[i[0]]]
        item.extend(list(temp_df.drop_duplicates('Book-Title')['Book-Title'].values))
        item.extend(list(temp_df.drop_duplicates('Book-Title')['Book-Author'].values))
        item.extend(list(temp_df.drop_duplicates('Book-Title')['Image-URL-M'].values))
        
        data.append(item)
    
    return data
```
